```python
#Reads weaver word list
file_name = 'weaver_dict.txt'

with open(file_name, 'r') as file:
    weaver_list = file.read().splitlines()
```

```python
def possible_words(input_word: str) -> list:
    """Finds all valid words in the dictionary that differ by exactly one letter."""
    possible_word_list = []
    for word in weaver_list:
        if len(word) != len(input_word):
            continue
        same = 0
        i = 0
        for letter in input_word:
            if word[i] == input_word[i]:
                same = same + 1
            i = i + 1
        if same == len(input_word) - 1:
             possible_word_list.append(word)
    return possible_word_list

#Example Usage:
#print(possible_words("spot"))
#Output:
#['scot', 'shot', 'slot', 'snot', 'soot', 'spat', 'spit', 'stot', 'swot']
```

```python
from collections import deque
def path(start_word: str, end_word: str) -> list:
    """Given start and end word determines path by creating queues of [word,[path]] and looking through all possibilities."""
    queue = deque([[start_word, [start_word]]])
    visited = set([start_word])
    while queue:
        current_word, current_path = queue.popleft()
        neighbors = possible_words(current_word)
        if current_word == end_word:
            return current_path
        for word in neighbors:
            if word not in visited:
                visited.add(word)
                queue.append([word, current_path + [word]])
    return "No Path Found"

#Example Usage:
#print(path('shot', 'boat'))
#Output:
#['shot', 'soot', 'boot', 'boat']
```
